```python
import import_ipynb
from expense import Expense
import datetime
import calendar
```

```python
def main():
    print(f"» Running Expense Tracker!")
    expense_file_path = "expenses.csv"
    budget = 4000
    
    # Get user input for expense.
    expense = get_user_expense()
    # print(expense)
    
    # Write their expense into a file.
    save_expenses_to_file(expense, expense_file_path)
    
    # Read file and summerize expenses.
    summerize_expenses(expense_file_path, budget)
```

```python
def get_user_expense():
    print(f"» Getting User Expense!")
    expense_name = input("Enter Expense name: ")
    expense_amount = float(input("Enter Expense amount: "))
    print(f"You've entered {expense_name}, {expense_amount}")
    
    expense_categories = [
        "Food", "Home", "Work", "Fun", "Misc"
    ]
    
    while True:
        print("Select a category: ")
        for i, category_name in enumerate(expense_categories):
            print(f"   {i+1}. {category_name}")
            
        no_of_categories = len(expense_categories)
        
        value_range = f"[1 - {no_of_categories}]"
        try:
            selected_index = int(input(f"Enter a category number: {value_range}")) - 1
        
        except ValueError:
            print("Enter an Integer value!")
            continue
            
        if 0 <= selected_index < no_of_categories:
            selected_category = expense_categories[selected_index]
            new_expense = Expense(
                name=expense_name, category=selected_category,
                amount=expense_amount 
            )
            return new_expense
        else:
            print("Enter valid category number!")
            
            
```

```python
def save_expenses_to_file(expense: Expense, expense_file_path):
    print(f"» Saving Expenses to a File: {expense} to {expense_file_path}")
    with open(expense_file_path, "a") as f:
        f.write(f"{expense.name},{expense.category},{expense.amount}\n")
        
```

```python
def summerize_expenses(expense_file_path, budget):
    print(f"» Summerizing Expenses!")
    expenses: list[Expense] = []
    with open(expense_file_path, "r") as f:
        lines = f.readlines()
        for line in lines:
            #stripped_line = line.strip() # to strip leading and traling white space and new line
            expense_name, expense_category, expense_amount = line.strip().split(",")
            #print(f"Name: {expense_name}; Category: {expense_category}; Amount: ${expense_amount}")
            line_expense = Expense(
                name=expense_name, category=expense_category,
                amount=float(expense_amount) 
            )
            #print(line_expense)
            expenses.append(line_expense)
        #print(expenses)
        
    amount_by_category = {}
    #total_expense = 0
    for expense in expenses:
        #total_expense += expense.amount
        key = expense.category
        # amount_by_category[key] = 0.0
        if key in amount_by_category:
            amount_by_category[key] += expense.amount
        else:
            amount_by_category[key] = expense.amount
    
    total_expense = sum([ex.amount for ex in expenses])
    remaining_budget = budget - total_expense
            
    print("Expenses by Category")
    for key, amount in amount_by_category.items():
        print(f"  »{key}: ${amount:.2f}")
    print(f"» Total expense = ${total_expense:.2f}")
    print(f"» Reaming Budget = ${remaining_budget:.2f}")
    
    now = datetime.datetime.now()
    days_in_month = calendar.monthrange(now.year, now.month)[1]
    remaining_days = days_in_month - now.day
    print(f"» Remaining days in the current month = ${remaining_days}")
    
    daily_budget = remaining_budget/remaining_days
    print(f"» Suggested daily budget = ${daily_budget:.2f}")
```

```python
# to make sure main() is run only from this file, not when the file imported by another file/script
if __name__ == "__main__": 
    main()
```
